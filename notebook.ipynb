```python
import pandas as pd

intakes = pd.read_csv("aac_intakes.csv")
outcomes = pd.read_csv("aac_outcomes.csv")
```

```python
intakes.head()
```

```python
outcomes.head()
```

```python
df = pd.merge(intakes, outcomes, on="animal_id", how="inner")
df
```

```python
df.head()
```

```python
df.shape
```

```python
df = df[[
    'animal_id',
    'animal_type_x',
    'breed_x',
    'color_x',
    'datetime_x',
    'datetime_y',
    'outcome_type',
    'sex_upon_intake',
    'age_upon_intake'
]]
df
```

```python
df = df.rename(columns={
    'animal_id': 'AnimalID',
    'animal_type_x': 'AnimalType',
    'breed_x': 'Breed',
    'color_x': 'Color',
    'datetime_x': 'EntryDate',
    'datetime_y': 'ReleaseDate',
    'outcome_type': 'OutcomeType',
    'sex_upon_intake': 'Sex',
    'age_upon_intake': 'Age'
})
```

```python
df
```

```python
df['EntryDate'] = pd.to_datetime(df['EntryDate'])
df['ReleaseDate'] = pd.to_datetime(df['ReleaseDate'])
```

```python
df.info()
```

```python
df['Year'] = df['ReleaseDate'].dt.year
df['Month'] = df['ReleaseDate'].dt.month
```

```python
df.head()
```

```python
df['IsAdopted'] = df['OutcomeType'] == 'Adoption'
```

```python
df[['OutcomeType', 'IsAdopted']].head(10)
```

```python
df.isnull().sum()
```

```python
df = df.dropna(subset=['AnimalType', 'OutcomeType', 'ReleaseDate'])
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.to_csv("final_data.csv", index=False)
```
